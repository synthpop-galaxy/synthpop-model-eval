```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
dat = pd.read_csv('mulens_rates_vvv_src17.txt')
dat2 = pd.read_csv('mulens_rates_vvv_NONSD_mag17.txt')
dat
```

```python
b_bins = np.array([-5,-1,1,5])
l_bins = np.array([-10,-5,0,5,10])
```

```python
for i in range(len(l_bins)-1):
    for j in range(len(b_bins)-1):
        subset = dat[(dat['l']>l_bins[i]) & (dat['l']<l_bins[i+1])
                     & (dat['b']>b_bins[j]) & (dat['b']<b_bins[j+1])]
        avg_logt = np.sum(subset['avg_logt']*subset['eventrate_area'])/np.sum(subset['eventrate_area'])
        avg_t = np.sum(subset['avg_t']*subset['eventrate_area'])/np.sum(subset['eventrate_area'])
        print(10**avg_logt, avg_t)
```

```python
for i in range(len(l_bins)-1):
    for j in range(len(b_bins)-1):
        subset = dat[(dat['l']>l_bins[i]) & (dat['l']<l_bins[i+1])
                     & (dat['b']>b_bins[j]) & (dat['b']<b_bins[j+1])]
        avg_evrate = np.sum(subset['eventrate_area'])/len(subset['eventrate_area'])
        print(avg_evrate)
```

```python
cm = plt.get_cmap('viridis')
cm_min,cm_max = 0,240
fig,ax=plt.subplots()
plt.vlines(l_bins,-10,10, color='gray',linestyle='--')
plt.hlines(b_bins,-10,10, color='gray',linestyle='--')

for i in range(len(l_bins)-1):
    for j in range(len(b_bins)-1):
        subset = dat[(dat['l']>l_bins[i]) & (dat['l']<l_bins[i+1])
                     & (dat['b']>b_bins[j]) & (dat['b']<b_bins[j+1])]
        avg_evrate = np.sum(subset['eventrate_area'])/len(subset['eventrate_area'])

        ptc = mpl.patches.Rectangle((l_bins[i],b_bins[j]), l_bins[i+1]-l_bins[i], b_bins[j+1]-b_bins[j],
                                    color=cm((avg_evrate-cm_min)/(cm_max-cm_min)),zorder=1)
        plt.text((l_bins[i]+l_bins[i+1])/2,(b_bins[j]+b_bins[j+1])/2,str(np.round(avg_evrate)),
                 va='center',ha='center',color='white',zorder=3,fontsize=16)
        ax.add_patch(ptc)
        #cax = plt.add_axes([0.92, 0.15, 0.02, 0.7]) # [left, bottom, width, height]
ax.invert_xaxis()
plt.colorbar(mpl.cm.ScalarMappable(norm=mpl.colors.Normalize(vmin=cm_min,vmax=cm_max),cmap=cm), 
                     label=r'$\Gamma$', ax=ax) #, cax=cax)
```

```python
b_bins = np.arange(-4,4.1,0.25)
#l_bins = np.array([-10,-5,0,5,10])
lmin,lmax = -1.25,0.26 #-4.0,4.0
def get_means(dat):
    n_star = []
    ev_area = []; ev_src=[]
    for j in range(len(b_bins)-1):
        subset = dat[(dat['l']>lmin) & (dat['l']<lmax)
                     & (dat['b']>b_bins[j]) & (dat['b']<b_bins[j+1])]
        n_star.append(np.mean(subset['n_source']/subset['sa_source']))
        ev_area.append(np.mean(subset['eventrate_area']))
        ev_src.append(np.mean(subset['eventrate_source']*subset['n_source']/subset['sa_source']
                             )/np.mean(subset['n_source']/subset['sa_source']))
    return n_star, ev_area, ev_src
n_star, ev_area, ev_src = get_means(dat)
n_star2, ev_area2, ev_src2 = get_means(dat2)
```
